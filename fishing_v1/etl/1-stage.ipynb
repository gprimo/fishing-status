```python
import os

import pyspark.sql.functions as F
import pyspark.sql.types as T
```

```python
from pathlib import Path
```

```python
CATALOGO = 'fishing'
VOLUME = 'raw'
BASE_PATH = f'/Volumes/{CATALOGO}/{VOLUME}/fishing'
```

```python
lista_arquivos = [os.path.join(BASE_PATH, file) for file in os.listdir(BASE_PATH) if file.endswith('.csv')]
lista_arquivos
```

```python
def read_csv(file_path):
  return spark.read.csv(file_path, header=True, inferSchema=True)
```

```python
fishing_fixed_gear_df = read_csv(lista_arquivos[0])
fishing_pole_and_line_df = read_csv(lista_arquivos[1])
fishing_purse_seines_df = read_csv(lista_arquivos[2])
```

```python
fishing_fixed_gear_df = fishing_fixed_gear_df.withColumn(
    "timestamp",
    F.from_unixtime(F.col("timestamp")).cast("timestamp")
).withColumn(
    'mmsi',
    F.col('mmsi').cast('long').cast('string'))
fishing_pole_and_line_df = fishing_pole_and_line_df.withColumn(
    "timestamp",
    F.from_unixtime(F.col("timestamp")).cast("timestamp")
).withColumn(
    'mmsi',
    F.col('mmsi').cast('long').cast('string'))

fishing_purse_seines_df = fishing_purse_seines_df.withColumn(
    "timestamp",
    F.from_unixtime(F.col("timestamp")).cast("timestamp")
).withColumn(
    'mmsi',
    F.col('mmsi').cast('long').cast('string'))

```

```python
fishing_fixed_gear_nm_df = (
    fishing_fixed_gear_df
    .filter(
        (F.col('is_fishing') > 0 )
    )
    .select(
        'mmsi',
        'timestamp',
        'lat',
        'lon',
        'is_fishing'
    )
    .orderBy(F.col("mmsi").desc())
)

fishing_pole_and_line_nm_df = (
    fishing_pole_and_line_df
    .filter(
        (F.col('is_fishing') > 0 )
    )
    .select(
        'mmsi',
        'timestamp',
        'lat',
        'lon',
        'is_fishing'
    )
    .orderBy(F.col("mmsi").desc())
)

fishing_purse_seines_df_nm_df = (
    fishing_purse_seines_df
    .filter(
        (F.col('is_fishing') > 0 )
    )
    .select(
        'mmsi',
        'timestamp',
        'lat',
        'lon',
        'is_fishing'
    )
    .orderBy(F.col("mmsi").desc())
)
```

```python
OUTPUT_PATH = Path(BASE_PATH, 'parquet')
OUTPUT_PATH.mkdir(exist_ok=True)

(
fishing_fixed_gear_nm_df
.write
.mode('overwrite')
.parquet(str(Path(OUTPUT_PATH, 'fishing_fixed_gear_nm_df')))
)
```

```python
OUTPUT_PATH = Path(BASE_PATH, 'parquet')
OUTPUT_PATH.mkdir(exist_ok=True)

(
fishing_pole_and_line_nm_df
.write
.mode('overwrite')
.parquet(str(Path(OUTPUT_PATH, 'fishing_pole_and_line_nm_df')))
)
```

```python
OUTPUT_PATH = Path(BASE_PATH, 'parquet')
OUTPUT_PATH.mkdir(exist_ok=True)

(
fishing_purse_seines_df_nm_df
.write
.mode('overwrite')
.parquet(str(Path(OUTPUT_PATH, 'fishing_purse_seines_df_nm_df')))
)
```
